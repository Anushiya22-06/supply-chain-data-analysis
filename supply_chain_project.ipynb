```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))
sns.histplot(sp['Sales'], bins=30, kde=True)

plt.title('Distribution of Sales')
plt.xlabel('Sales')
plt.ylabel('Number of Transactions')
plt.show()# it shows x axis is amount of money and y axis shows no of transaction how many people have spent the amount
```

```python
plt.figure(figsize=(10, 6))

category_sales.head(10).sort_values().plot(kind='barh')

plt.title('Top 10 Categories by Sales')
plt.xlabel('Total Sales')
plt.ylabel('Category Name')
plt.savefig('top_10_categories_by_sales.png', dpi=300, bbox_inches='tight')
plt.show()# it shows x ais = howmuch sales amount happened and y axis is category names of top 10
# it shows how much money it makse by category
```

```python
order_delivery_status = sp.groupby('Delivery Status')['Order Id'].nunique()
order_delivery_status# it shows the no of times deleivery status apperared uniquely for each unique id satus
```

```python
shipping_delivery = pd.crosstab(sp['Shipping Mode'],sp['Delivery Status'])

shipping_delivery# each shipping mode category and its delivery status how many of them do it corretly
```

```python
shipping_delivery_orders = sp.groupby(
    ['Shipping Mode', 'Delivery Status']
)['Order Id'].nunique()

shipping_delivery_orders
```

```python
shipping_mode_total = shipping_delivery_orders.groupby(level=0).sum()
shipping_mode_total# it shows the total no of shipping mode for each category
#it counts which type of deleivery sttuts is it it dosen't consider it
```

```python
shipping_delivery_percentage = (
    shipping_delivery_orders
    .div(shipping_mode_total, level='Shipping Mode')
    * 100
)

shipping_delivery_percentage
```

```python
shipping_delivery_graph = shipping_delivery_percentage.unstack(fill_value=0)

shipping_delivery_graph
```

```python
shipping_delivery_graph.plot(kind='bar',stacked=True,figsize=(10, 6))

plt.title('Delivery Status by Shipping Mode')
plt.xlabel('Shipping Mode')
plt.ylabel('Percentage of Unique Orders (%)')
plt.xticks(rotation=0)
plt.legend(title='Delivery Status')
plt.tight_layout()

plt.savefig('delivery_status_by_shipping_mode.png',dpi=300,bbox_inches='tight')

plt.show()
```

```python
sp['Ship_Diff'] = (
    sp['Days for shipping (real)']
    - sp['Days for shipment (scheduled)'])

sp['Ship_Diff'].value_counts().sort_index()#-2 -- 2 days early ,0- correct day,1-one day late likewise
```

```python
#early shipping and its no orders
ship_diff_count = sp['Ship_Diff'].value_counts().sort_index()
ship_diff_count.plot(kind='bar',figsize=(9, 5))
plt.title('Shipping Difference')
plt.xlabel('Shipping Difference (Days)')
plt.ylabel('Number of Orders')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('shipping_difference.png',dpi=300,bbox_inches='tight')
plt.show()# it show what days sdelivery shipping happens within how many no of orders
```

```python
import unicodedata

for col in sp.select_dtypes(include='object').columns:
    sp[col] = sp[col].apply(
        lambda x: ''.join(
            c for c in unicodedata.normalize('NFD', str(x))
            if unicodedata.category(c) != 'Mn'
        ) if x is not None else x
    )
```

```python
region_sales = sp.groupby('Order Region')['Sales'].sum().sort_values(ascending=False)# it shows each region total sales amount
pd.set_option('display.float_format', '{:,.2f}'.format)# helps to show full real world songs ..
region_sales
```

```python
region_profit = (sp.groupby('Order Region')['Order Profit Per Order'].sum().sort_values(ascending=False))
region_profit# each region how much total profit happens
```

```python
segment_sales = (sp.groupby('Customer Segment')['Sales'].sum().sort_values(ascending=False))
segment_sales
```

Consumer → normal individual customers buying for personal use
Corporate → businesses/companies buying products
Home Office → people buying products for their home-office/business use

```python
plt.figure(figsize=(10, 8))
region_sales.sort_values().plot(kind='barh')
plt.title('Sales by Region')
plt.xlabel('Total Sales')
plt.ylabel('Order Region')
plt.tight_layout()
plt.savefig('sales_by_region.png',dpi=300, bbox_inches='tight')
plt.show()
```

```python
region_segment_sales = (sp.groupby(['Order Region', 'Customer Segment'])['Sales'].sum())
region_segment_sales
```

```python
region_segment_percentage = (region_segment_sales.div(region_sales, level='Order Region')* 100)
region_segment_percentage# for each oder region category by customer segment show percentage sales
```

```python
region_segment_graph = region_segment_percentage.unstack()

region_segment_graph.plot(
    kind='bar',stacked=True,figsize=(12, 7))
plt.title('Customer Segment Sales Mix by Region')
plt.xlabel('Order Region')
plt.ylabel('Percentage of Regional Sales (%)')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Customer Segment')
plt.tight_layout()
plt.savefig('customer_segment_sales_mix_by_region.png',
    dpi=300,
    bbox_inches='tight')
plt.show()
```
